## 一、字符串的乐趣

```python
def is_palindrome(s):
    temp = ""
    for char in s:
        if char != " ":
            temp = temp + char
    lower_str = temp.lower()
    return lower_str == lower_str[::-1]


def is_abecedarian(s):
    no_space = ""
    for c in s:
        if c != " ":
            no_space += c
    low = no_space.lower()
    for i in range(len(low)-1):
        if low[i] > low[i+1]:
            return False
    return True


def remove_vowels(s):
    vowels = {'a','e','i','o','u'}
    res = ""
    for char in s:
        if char.lower() not in vowels:
            res += char
    return res

# 测试回文
print(is_palindrome("level"))
print(is_palindrome("T A C O cat"))
print(is_palindrome("hello"))

# 测试字母顺序
print(is_abecedarian("adder"))
print(is_abecedarian("beet"))
print(is_abecedarian("abcd efgh xyz"))

# 删除元音测试
print(remove_vowels('cat'))
print(remove_vowels('audio'))
print(remove_vowels("AEIOU"))
```

## 二、列表的乐趣

```python
def list_reverse(input_list):
    if not isinstance(input_list, list):
        raise TypeError("输入必须是列表")
    new_list = []
    idx = len(input_list)-1
    while idx >= 0:
        new_list.append(input_list[idx])
        idx -= 1
    return new_list


def is_sorted(seq):
    if len(seq) <= 1:
        return True
    for i in range(len(seq)-1):
        if seq[i] > seq[i+1]:
            return False
    return True

print(list_reverse([1,2,3]))
print(is_sorted([1,2,3,4]))
print(is_sorted([1,3,2]))
print(is_sorted("abcde"))
```

## 三、作为向量的元组

```python
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    if not isinstance(s, (int, float)):
        raise TypeError("标量s必须是整数或者浮点数")
    for item in t:
        if not isinstance(item, (int, float)):
            raise TypeError("元组内元素必须是整数或浮点数")
    out = []
    for num in t:
        out.append(num * s)
    return tuple(out)


def vec_inner_product(t1, t2):
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("输入必须是元组")
    if len(t1) != len(t2):
        raise ValueError("两个向量元组长度要相同")
    total = 0.0
    for a,b in zip(t1,t2):
        total = total + a * b
    return total


def check_valid_mx(mx):

    if not isinstance(mx, tuple):
        return False
    # 外层每一个都得是元组
    for row in mx:
        if not isinstance(row, tuple):
            return False

    row_len = len(mx[0])
    for row in mx:
        if len(row) != row_len:
            return False
        
        for elem in row:
            if not isinstance(elem, (int, float)):
                return False
    return True


def mx_vec_mult(matrix, vec):
    if not check_valid_mx(matrix):
        raise ValueError("矩阵格式不合法")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    n_col = len(matrix[0])
    if len(vec) != n_col:
        raise ValueError("矩阵列数和向量维度不一致")
    result = []
    for row in matrix:
        dot = 0
        for m_elem, v_elem in zip(row, vec):
            dot += m_elem * v_elem
        result.append(dot)
    return tuple(result)

print(vec_scalar_mult((1,2,3),2))
print(vec_inner_product((1,2),(3,4))) #1*3+2*4 =11.0

mat = ((1,2),(3,4))
print(check_valid_mx(mat))
print(mx_vec_mult(mat, (1,1)))
```

## 四、更多向量的乐趣

```python
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int) or dim < 0:
            raise ValueError("维度必须是非负整数")

        if elements is None:
            elements = [0] * dim
            
        if not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须是列表或元组")
            
        if len(elements) != dim:
            raise ValueError("维度和元素个数不一致")
            
        for val in elements:
            if not isinstance(val, (int, float)):
                raise TypeError("元素必须是数字")

        self.dim = dim
        self.elements = tuple(elements)

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return False
        if self.dim != other.dim:
            raise ValueError("维度不同，无法比较")
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和向量比较")
        if self.dim != other.dim:
            raise ValueError("维度不同，无法比较")
        return self.elements < other.elements

    def __le__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和向量比较")
        if self.dim != other.dim:
            raise ValueError("维度不同，无法比较")
        return self.elements <= other.elements

    def __gt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和向量比较")
        if self.dim != other.dim:
            raise ValueError("维度不同，无法比较")
        return self.elements > other.elements

    def __ge__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和向量比较")
        if self.dim != other.dim:
            raise ValueError("维度不同，无法比较")
        return self.elements >= other.elements

    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("必须是 Vector 对象")
        if self.dim != other.dim:
            raise ValueError("维度不同，无法计算内积")
        
        total = 0
        for i in range(self.dim):
            total += self.elements[i] * other.elements[i]
        return float(total)

    def __mul__(self, other):
        if isinstance(other, (int, float)):
            new_elements = []
            for val in self.elements:
                new_elements.append(val * other)
            return Vector(self.dim, new_elements)
            
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("维度不同，无法逐元素相乘")
            new_elements = []
            for i in range(self.dim):
                new_elements.append(self.elements[i] * other.elements[i])
            return Vector(self.dim, new_elements)
        else:
            raise TypeError("乘数必须是数字或 Vector 对象")

    def __rmul__(self, other):
        return self.__mul__(other)


# 测试
print("--- 测试 Vector 类 ---")
v1 = Vector(2, (2, 3))
v2 = Vector(2, (4, 5))

print(v1 == v2)
print(v1 < v2)
print(v1.dot(v2))

# 测试乘法
print((v1 * v2).elements)
print( (v1 * 2).elements)
print((2 * v1).elements)
```
