## 实验1：正则表达式的基础匹配

```python
import re

text1 = "cat on mat; later: cats and dogs"

# 匹配数字
print("匹配数字:", re.findall(r"\d+", text1))

# 匹配空白字符
print("匹配空白:", re.findall(r"\s+", text1))

# 匹配独立的单词 cat（前后有边界，不会匹配到cats里的cat）
print("匹配独立cat:", re.findall(r"\bcat\b", text1))
```

```python
text2 = "cat dog bird CAT"
print("cat|dog:", re.findall(r"cat|dog", text2))

print("a|aa|aaa:", re.findall(r"a|aa|aaa", "aaaa"))
print("aaa|aa|a:", re.findall(r"aaa|aa|a", "aaaa"))
```

```python
text3 = "ct cat caat caaat"

print("ca*t:", re.findall(r"ca*t", text3))
print("ca+t:", re.findall(r"ca+t", text3))
print("ca?t:", re.findall(r"ca?t", text3))
print("ca{2}t:", re.findall(r"ca{2}t", text3))
print("ca{1,2}t:", re.findall(r"ca{1,2}t", text3))
```

- `*`：表示前面的字符出现0次或多次。
- `+`：表示前面的字符出现1次或多次（至少1次）。
- `?`：表示前面的字符出现0次或1次（可有可无）。
- `{2}`：表示前面的字符必须严格出现2次。
- `{1,2}`：表示前面的字符出现1到2次。

## 实验2：分组、回溯引用与替换

```python
text_email = "联系: alice@example.com; 备选: bob@test.org"

pattern = r"(\w+)@([\w.]+)"
matches = re.findall(pattern, text_email)

for user, domain in matches:
    print(f"用户名: {user}, 域名: {domain}")
```

```python
text_dup = "cat cat, dog cat, bird bird"

pattern = r"\b(\w+)\s+\1\b"
print("重复的单词:", re.findall(pattern, text_dup))
```

```python
text_date = "2024-05-01\n2024/05/02\n2024.05.03"

new_text = re.sub(r"(\d{4})[-/.](\d{2})[-/.](\d{2})", r"\2-\3-\1", text_date)

print("转换前:\n" + text_date)
print("转换后:\n" + new_text)
```

```python
text_phone = "联系我: 123-456-7890 或 987.654.3210"

new_phone = re.sub(r"(\d{3})[-.](\d{3})[-.](\d{4})", r"(\1) \2-\3", text_phone)

print(new_phone)
```

## 实验3：使用 BeautifulSoup 解析 HTML

```python
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲: 正则表达式</p>
<p class="course">第七讲: 结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""

soup = BeautifulSoup(html_doc, "html.parser")
```

```python
print("页面标题文本:", soup.title.text)

h1_tag = soup.h1
print("h1的标签名:", h1_tag.name)
print("h1的文本:", h1_tag.text)
```

```python
courses = soup.find_all("p", class_="course")
print("--- 课程段落 ---")
for c in courses:
    print(c.text)

links = soup.find_all("a")
print("--- 链接 ---")
for link in links:
    print("文字:", link.text, "| 链接:", link["href"])
```

```python
print("--- 整个文档的文本 ---")
print(soup.get_text())

print("--- 单个段落的文本 ---")
print(courses[0].get_text())

print("--- 单个链接的文本 ---")
print(links[0].get_text())
```

- `soup.get_text()` 把整个 HTML 里所有的换行、文本都抓出来了，包含标题、段落、链接文字，是一大段混杂的文本。
- 单个段落（`courses[0]`）和单个链接（`links[0]`）的 `get_text()` 只会提取对应标签内部的文字，很干净。
所以在做网页爬虫提取具体数据时，通常先用 `find_all` 定位到具体的标签，再用 `get_text()` 提取纯文本，而不是直接抓整个页面，那样数据太乱了不好处理。
